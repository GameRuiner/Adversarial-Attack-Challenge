```python
import os
from PIL import Image
import matplotlib.pyplot as plt
```

```python
base_path = '../AdvCelebA'
label_file = os.path.join(base_path, 'attack_CelebA.txt')
image_dir = os.path.join(base_path, 'images')
```

```python
clean_images = []
adv_images = []

with open(label_file, 'r') as f:
    lines = f.readlines()

for line in lines:
    filename, label = line.strip().split()
    if label == '0' and len(clean_images) < 5:
        clean_images.append(filename)
    elif label == '1' and len(adv_images) < 5:
        adv_images.append(filename)
    if len(clean_images) == 5 and len(adv_images) == 5:
        break
```

```python
def show_images(image_list, title):
    plt.figure(figsize=(15, 3))
    for i, filename in enumerate(image_list):
        img_path = os.path.join(image_dir, filename)
        image = Image.open(img_path)
        plt.subplot(1, 5, i + 1)
        plt.imshow(image)
        plt.title(filename)
        plt.axis('off')
    plt.suptitle(title)
    plt.show()

show_images(clean_images, "Clean Images")
show_images(adv_images, "Adversarial Images")
```
